## Part 2: Import the required libraries

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_vector, plot_bloch_multivector
```

## Part 3: Visualizing basis states

A Bloch vector is written as **[x, y, z]** in Cartesian coordinates.

- $|0\rangle$ is at the **north pole**
- $|1\rangle$ is at the **south pole**

### Task 1
Run the code below and observe the plots.


```python
plot_bloch_vector([0, 0, 1], title="State |0>")
```

```python
plot_bloch_vector([0, 0, -1], title="State |1>")
```

### Reflection
1. Why is $|0\rangle$ at the top of the sphere?
2. Why is $|1\rangle$ at the bottom?
3. What does the **z-axis** represent here?


### Write your reflections for Task 1 here.
1. In the Bloch sphere representation, the state vector is defined by the angle $\theta$ from the positive z-axis. For the state $|0\rangle$, the angle $\theta = 0$. Since a $0^{\circ}$ angle from the vertical axis points straight up, $|0\rangle$ is mapped to the North Pole  
2. Following the mathematical mapping $|\psi\rangle = \cos(\frac{\theta}{2})|0\rangle + e^{i\phi}\sin(\frac{\theta}{2})|1\rangle$, the state $|1\rangle$ occurs when $\theta = \pi$ ($180^{\circ}$). This points the vector in the opposite direction of $|0\rangle$, which is the South Pole  
3. The z-axis represents the Computational Basis (also known as the Standard Basis). Its endpoints correspond to the two classical states of a bit, $0$ and $1$, which are the outcomes of a standard measurement

## Part 5: Using spherical coordinates

Instead of Cartesian coordinates, the Bloch Sphere can also be described using spherical coordinates:

- $r$ = radius
- $\theta$ = angle from the positive z-axis
- $\phi$ = angle around the z-axis

### Task 3
Plot a state using spherical coordinates.


```python
plot_bloch_vector([1, np.pi/2, np.pi/3], coord_type='spherical')
```

### Reflection
1. What happens when $\theta = \pi/2$?
2. What role does $\phi$ play?
3. Why is the radius equal to 1 for a pure state?


### Write your reflections for Task 3 here.
1. When $\theta = \pi/2$ (or $90^\circ$), the state vector lies on the equator of the Bloch sphere. This means the qubit is in an equal superposition, where there is a 50% probability of measuring 0 and a 50% probability of measuring 1  
2. $\phi$ represents the relative phase of the quantum state. While $\theta$ determines the probability of the outcome, $\phi$ determines the "direction" the state points on the equator (the rotation around the z-axis). It is crucial for quantum interference  
3. The radius $r = 1$ represents the normalization of the quantum state. In quantum mechanics, the total probability of all possible outcomes must equal 1 (100%). Any state vector that reaches the surface of the sphere ($r=1$) is a pure state, meaning we have complete information about the system   

## Part 6: Convert spherical coordinates to Cartesian coordinates

### Task 4
Run the code below to convert spherical coordinates into Cartesian coordinates.


```python
# Define the spherical coordinates
r = 1
#theta = np.pi/2
#theta = 0
theta = np.pi
phi = np.pi/3

# Convert spherical to Cartesian coordinates
x = r * np.sin(theta) * np.cos(phi)
y = r * np.sin(theta) * np.sin(phi)
z = r * np.cos(theta)

print("x =", x)
print("y =", y)
print("z =", z)
```

```python
plot_bloch_vector([x, y, z], title="Converted Cartesian point")
```

### Reflection
1. How do the Cartesian coordinates relate to the plotted point?
2. If you change $\theta$ to 0, where will the point move?
3. If you change $\theta$ to $\pi$, where will the point move?


### Write your reflections for Task 4 here.
1. The Cartesian coordinates $[x, y, z]$ define the exact position of the arrow's tip in 3D space. The $z$ value determines how far the point is from the equator toward the poles, while $x$ and $y$ determine its orientation and distance within the horizontal plane  
2. The point will move to the North Pole ($|0\rangle$ state). Mathematically, when $\theta = 0$, the formula $z = r \cos(0)$ equals 1, and $x = r \sin(0) \cos(\phi)$ equals 0, placing the vector at [0, 0, 1]  
3. The point will move to the South Pole ($|1\rangle$ state). At $\theta = \pi$ ($180^\circ$), the formula $z = r \cos(\pi)$ equals -1, resulting in the coordinates [0, 0, -1]

## Part 8: Apply the Hadamard gate

The **Hadamard gate** takes $|0\rangle$ into a superposition state.

### Task 5
Before running the cell:
- Predict whether the final state will be at a pole or on the equator.


```python
qc = QuantumCircuit(1)
qc.h(0)

print(qc)

state = Statevector.from_instruction(qc)
plot_bloch_multivector(state)
```

### Reflection
1. Where did the state move?
2. Why is the state no longer at the north pole?
3. What does this tell you about superposition?


### Write your reflections for Task 5 here.
1. The state moved from the North Pole ($|0\rangle$) to the positive x-axis ($|+\rangle$ state) on the equator  
2. The Hadamard gate performs a rotation that mixes the basis states $|0\rangle$ and $|1\rangle$. Geometrically, it rotates the vector around the $[1, 0, 1]$ axis, effectively pushing the state from the vertical Z-axis down to the horizontal plane 
3. It tells us that superposition is a geometric middle ground between classical 0 and 1. In this state, the qubit is not "either 0 or 1," but exists in a state that has a 50% probability of collapsing into either value upon measurement

The **X gate** flips $|0\rangle$ to $|1\rangle$.

### Reflection

2. Why can the X gate be thought of as a bit-flip gate?


### Write your reflections for Task 6 here.

2. Because its effect on the computational basis states mirrors the classical NOT gate: it transforms $|0\rangle$ into $|1\rangle$ and $|1\rangle$ into $|0\rangle$. In the context of the Bloch sphere, this is a $180^\circ$ rotation that "flips" the qubit's orientation relative to the Z-axis 

## Part 10: Apply H followed by Z

The **Z gate** changes phase. This is important because phase changes are not always obvious from measurement probabilities alone, but they still affect the Bloch Sphere representation.

### Task 7
Predict how the state will move after applying:
1. H
2. then Z


```python
qc = QuantumCircuit(1)
qc.h(0)
qc.z(0)

print(qc)

state = Statevector.from_instruction(qc)
plot_bloch_multivector(state)
```

### Reflection
1. Did the point remain on the equator?
2. What changed after the Z gate?
3. Why is phase important in quantum computing?


### Write your reflections for Task 7 here.
1. Yes, the point remained on the equator. Because a rotation around the Z-axis does not change the "height" (latitude) of the state vector, it stays halfway between the poles.
2. The relative phase of the qubit changed. Geometrically, the vector rotated $180^\circ$ around the Z-axis, moving from the $[1, 0, 0]$ position to the $[-1, 0, 0]$ position. Mathematically, the state changed from $|+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}$ to $|-\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}}$
3. Phase is essential for quantum interference. Even though $|+\rangle$ and $|-\rangle$ both have a 50/50 probability of being measured as 0 or 1, they behave differently when combined with other gates. Phase allows quantum algorithms to cancel out wrong answers and amplify correct ones through constructive and destructive interference
                                                                                                                                                                                                                         

## Part 11: Compare a few gates

### Task 8
Run the next cell and compare the effect of different gates on the Bloch Sphere.


```python
gate_sets = {
    "H": ["h"],
    "X": ["x"],
    "H then X": ["h", "x"],
    "H then Z": ["h", "z"],
    "H then S": ["h", "s"]
}

for name, gates in gate_sets.items():
    qc = QuantumCircuit(1)
    for gate in gates:
        getattr(qc, gate)(0)
    print(f"\n{name}")
    print(qc)
    state = Statevector.from_instruction(qc)
    display(plot_bloch_multivector(state))
```

### Reflection
Choose any **two** gate sequences above and explain:
- where the state ended up
- whether the move looked like a flip or a rotation
- what you learned from the comparison


### Write your reflections for Task 8 here.
H then S  
The H then S gate sequence demonstrates how quantum gates can manipulate quantum states in a non-classical way.
1. After the H gate moved the state to the positive x-axis ($|+\rangle$), the S gate moved it to the positive y-axis 
2. It looked like a rotation (specifically a $90^\circ$ rotation around the Z-axis) 
3. I learned that the S gate is a "Phase gate." Unlike the Z gate which rotates $180^\circ$, the S gate performs a smaller $90^\circ$ rotation, allowing for more precise control over the qubit's phase on the equator
