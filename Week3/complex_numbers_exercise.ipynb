# Complex Numbers Exercise

```python
import numpy as np
import matplotlib.pyplot as plt

# plot complex numbers
def plot_complex(z_list, labels, title):
    colors = ['blue', 'red', 'green', 'purple']
    plt.figure(figsize=(6, 6))
    for i, (z, label) in enumerate(zip(z_list, labels)):
        plt.arrow(0, 0, z.real, z.imag, head_width=0.2, head_length=0.2, fc=colors[i], ec=colors[i])
        plt.plot(z.real, z.imag, 'o', color=colors[i])
        plt.text(z.real, z.imag, ' '+label)
    plt.axhline(0, color='k')
    plt.axvline(0, color='k')
    plt.grid()
    plt.xlabel('Real')
    plt.ylabel('Imaginary')
    plt.title(title)
    plt.axis('equal')
    plt.show()
```

## 1. Take two complex numbers and visualize them

```python
z1 = 3 + 4j
z2 = 2 - 1j

print("z1 =", z1)
print("z2 =", z2)

plot_complex([z1, z2], ['z1', 'z2'], 'Complex Numbers')
```

## 2. Addition

```python
z_add = z1 + z2
print("z1 + z2 =", z_add)

plot_complex([z1, z2, z_add], ['z1', 'z2', 'z1+z2'], 'Addition')
```

## 3. Subtraction

```python
z_sub = z1 - z2
print("z1 - z2 =", z_sub)

plot_complex([z1, z2, z_sub], ['z1', 'z2', 'z1-z2'], 'Subtraction')
```
